```python
from sklearn.datasets import load_iris
from sklearn.tree import DecisionTreeClassifier
```

```python
iris=load_iris()
X=iris.data[:,2:]
y=iris.target
```

```python
tree_clf = DecisionTreeClassifier(max_depth=2)
tree_clf.fit(X, y)
```

```python
from sklearn import tree
import matplotlib.pyplot as plt
tree.plot_tree(tree_clf,feature_names=iris.feature_names[2:], class_names=iris.target_names,filled=True, rounded=True)
plt.show()
```
